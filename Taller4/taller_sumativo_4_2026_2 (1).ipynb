**Respuesta:**

Evaluamos el dataset con los seis principios del Human Rights-Based Approach to Data (OHCHR, 2018). Como es sintético y de uso educativo (Fatima, 2026), evaluamos su **diseño**: qué pasaría si un registro real se construyera de la misma forma.

| Principio | ¿Cumple? | Justificación |
|---|---|---|
| **Participación** | No | Las pacientes no participaron en definir qué se mide ni para qué. Lo diseñó una sola autora con fines de *machine learning*. |
| **Desagregación** | Parcial | Permite desagregar por edad, género e ingreso, pero no por territorio, etnia ni previsión, que es donde suelen aparecer las brechas de acceso. Como vimos en el Taller 2, sin la variable no se puede auditar el sesgo. |
| **Autoidentificación** | No | `Gender` solo tiene dos categorías asignadas por el algoritmo. Además, el 77 % de los hombres responde `Breastfeeding_History` (Taller 1): las categorías no se pensaron desde las personas. |
| **Transparencia** | Parcial | Declara que es sintético y publica un diccionario, pero no explica cómo se generó ni por qué hay exactamente 300 nulos en cinco variables. |
| **Privacidad** | Sí, por ser sintético | No expone a nadie real. Pero con `Patient_ID`, `Genetic_Mutation` e ingreso, un dataset real con esta estructura sería muy re-identificable. |
| **Rendición de cuentas** | No | No hay una institución responsable ni un canal para reclamar o corregir datos. |

**Conclusión:** el dataset **no cumple**. Solo cumple la privacidad, y únicamente porque nadie real está detrás. Por eso proponemos un marco de recolección para un registro real, basado en el HRBAD, el **consentimiento dinámico** (Kaye et al., 2015) y el modelo **Five Safes** (Desai et al., 2016):

1. **Participación:** un comité de pacientes, agrupaciones oncológicas, personal clínico y especialistas en ética define qué se recolecta y revisa los resultados antes de publicarlos.
2. **Desagregación y autoidentificación:** agregar territorio (región o zona urbana/rural), previsión y pueblo originario, publicados solo de forma agregada. El género y la etnia los informa la propia paciente, con opciones abiertas y de forma voluntaria.
3. **Consentimiento dinámico:** la paciente puede ver para qué se usan sus datos, aceptar o rechazar cada uso y retirar su consentimiento cuando quiera.
4. **Acceso por niveles (Five Safes):** los identificadores se separan y seudonimizan. Los datos reales solo los usan investigadores capacitados, en proyectos aprobados por un comité de ética, en entornos seguros y publicando solo resultados agregados. El dato genético nunca sale de ese entorno, y al público se entrega una versión sintética con privacidad diferencial.
5. **Rendición de cuentas:** una institución responsable, una metodología pública, un registro de quién accede y un canal para consultar, corregir o eliminar datos.

Elegimos este marco porque no se limita a proteger los datos: pone a las pacientes en el centro de las decisiones, que es la idea central del HRBAD.

**`Pregunta (4)` Proponga un método de recolección de este tipo de datos imaginando que fuese requerido para Chile. El método debe cumplir con el marco regulatorio chileno, considerando las políticas que entraron y entrarán en vigencia.**

Al momento de explicar el método propuesto, recuerda hacer alusión a principios, temas y/o artículos específicos de la regulación vigente o de discusiones académicas en torno a esta.

El método que proponemos es un registro oncológico de cáncer de mama con acceso por niveles, que cumple la Ley 19.628 y la Ley 20.584 (vigentes) y está diseñado para cumplir la Ley 21.719, que entra en vigencia próximamente. Lo opera un prestador de salud, junto con un equipo universitario de investigación. Los datos provienen de la atención clínica y después se reutilizan para el modelo, siguiendo estas etapas:

- **Base legal:** los datos clínicos van a la ficha clínica, que es información sensible y reservada y se conserva al menos 15 años (Ley 20.584, arts. 12 y 13). La notificación del caso al Registro Nacional del Cáncer tiene respaldo legal propio, y exclusivamente para ese fin (Ley 21.258). Usar los datos para investigar o entrenar un modelo es otro propósito y requiere consentimiento expreso, libre, informado y por escrito (Ley 20.584, art. 21; Ley 19.628, arts. 4 y 10; Ley 21.719, arts. 12 y 16).
- **Finalidad y minimización:** se recolecta solo lo necesario para cumplir el objetivo (Ley 21.719, art. 3). El `Patient_ID` se separa de los datos clínicos y se reemplaza por un seudónimo, y la edad y el ingreso se registran en tramos. `Smoking` y `Alcohol_Consumption` se responden de forma voluntaria.
- **Dato genético:** `Genetic_Mutation` requiere consentimiento aparte y específico, y se guarda solo en el entorno seguro. La ley reserva la información genética y prohíbe discriminar por ella (Ley 20.120, arts. 4 y 12), y los antecedentes oncológicos tienen límites de uso en seguros (Ley 21.258, art. 8° bis).
- **Transparencia y derechos:** la paciente recibe información clara sobre quién usa sus datos y para qué finalidad. Puede acceder, rectificar, suprimir, oponerse y retirar su consentimiento en cualquier momento. El prestador responde en 30 días corridos, prorrogables una vez (Ley 21.719, arts. 5 a 9, 11 y 12).
- **Seguridad y gobernanza:** se hace una evaluación de impacto, porque se trabaja con datos sensibles a gran escala. Se aplican cifrado y seudonimización, un registro de accesos y un protocolo de reporte de vulneraciones a la Agencia y a las pacientes afectadas. Un comité ético-científico aprueba el protocolo y se designa un delegado de protección de datos (Ley 21.719, arts. 14 quinquies, 14 sexies, 15 ter y 50; Ley 20.584, art. 22).
- **Publicación:** los resultados se publican de forma anonimizada (Ley 21.719, art. 16 bis). Lo que se entrega al público es la versión sintética con privacidad diferencial, y los datos reales no salen del entorno seguro.

**Participación de las pacientes.** El comité que revisa el protocolo incluye agrupaciones de pacientes oncológicas, algo que la propia Ley Nacional del Cáncer establece como principio (Ley 21.258, art. 2 letras b y c).

**Límites de la propuesta.** En primer lugar, la Agencia aún no está instalada, así que mientras tanto no existe el órgano que fiscalice estas obligaciones. En segundo lugar, un dataset sintético con privacidad diferencial no queda automáticamente anonimizado en términos legales, porque la Ley 21.719 exige que la anonimización sea irreversible (art. 2 letra k). Por eso se necesita evaluar el riesgo de reidentificación antes de publicarlo, y no basta con haber aplicado la técnica.

**`Pregunta (7)` Reporte los resultados de la técnica de anonimización (ModelInspector) y explique como se podría realizar activismo de datos con el conjunto anonimizado.**



**Respuesta:**

**El escenario.** Ahora el dataset trae el ID de cada paciente, el centro de salud donde se atiende y la comuna donde vive. Con esto se podría, por ejemplo, ver si en algunas comunas el cáncer se detecta más tarde que en otras.

**¿Por qué es un problema?** Aunque saquemos el nombre, una persona se puede reconocer juntando pocos datos. Sweeney (2000) mostró que el 87 % de la población de EE. UU. se puede identificar solo con su código postal, su género y su fecha de nacimiento, y Rocher et al. (2019) estimaron que con 15 datos demográficos se puede re-identificar al 99,98 % de las personas. En nuestro dataset pasa algo parecido: solo con la edad, el género y el ingreso, casi todas las pacientes quedan con una combinación única (9.997 de 10.000). Si además sabemos su comuna y su centro de salud, en una comuna pequeña sería muy fácil saber de quién se trata.

| | **Datos reales** | **Datos sintéticos** |
|---|---|---|
| **Pros** | Son exactos: sirven para atender a cada paciente y para analizar en detalle qué pasa en cada comuna. | Protegen a las pacientes y a sus familias, y se pueden compartir sin problema porque mantienen las tendencias generales. |
| **Contras** | Si se filtran, una paciente podría ser discriminada por su aseguradora o su empleador. En el caso de la mutación genética, también se expone a su familia, porque es hereditaria. | Son menos exactos en los grupos pequeños, que son justo los que más importa mirar. En el Taller 2 vimos que el modelo ya fallaba más en las pacientes jóvenes, y los hombres son solo el 7,2 % del dataset. Además, si se configuran mal (con un ε alto), pueden filtrar información. |

**¿Cuándo usar cada uno?** Los datos reales solo deberían verlos quienes realmente los necesitan: el equipo médico que atiende a la paciente y los investigadores autorizados que trabajen en un entorno seguro, publicando solo resultados generales (Desai et al., 2016). Para todo lo demás, como investigaciones abiertas, clases, probar modelos o hacer activismo de datos, basta con datos sintéticos, porque ahí interesa ver cómo se comportan los grupos y no quién es cada persona.

**¿Cómo lo resolveríamos?** Nosotros no publicaríamos los datos completos. Ser precisos y transparentes es importante, pero no justifica exponer la salud de las pacientes. Nuestra solución es simple: **los datos reales se quedan protegidos y lo que se comparte es una versión sintética**. En la práctica, esto significa:

1. Cambiar el ID de la paciente por un código que solo el centro de salud pueda asociar a la persona.
2. Antes de compartir, usar datos menos específicos: la región en vez de la comuna, tramos de edad en vez de la edad exacta y tramos de ingreso en vez del monto.
3. Publicar una versión sintética con privacidad diferencial, para que cualquiera pueda hacer análisis sin exponer a nadie.

Así el proyecto sigue siendo útil para detectar desigualdades entre comunas, pero sin que el costo lo paguen las pacientes. No tendría sentido buscar un acceso más justo a la salud poniendo en riesgo justamente a las personas más vulnerables.
