## ÍTEM 6: Desbalance de clases

Dentro de las variables demográficas identificadas o de las que establecieron como importantes, ¿existe algún desbalance de clases? A qué creen que se puede deber, expliquen las consecuencias que esto pudiese tener al momento de usar los datos en practicas como entrenamiento de modelos de machine learning. (Si es que no identifican desbalance de todas formas, expliquen cuáles son los riesgos de ese fenómeno en el contexto de su base de datos o campo de investigación correspondiente a los datos)

- En su respuesta citen/parafraseen al menos dos artículos sobre el impacto del desbalance de clases y/o la falta de representatividad de ciertas poblaciones (variables demográficas) en bases de datos y su uso posterior.

Uno de los obstáculos más comunes al trabajar con bases de datos clínicos complejos es que las categorías no están representadas en la misma proporción: suele haber muchos más casos de la condición "típica" que de la condición "rara" o menos frecuente (Gurcan & Soylu, 2024). En el caso de nuestra base de datos en específico, aunque la variable Edad no está desbalanceada, sí se observa un claro desbalance en las variables Cáncer y Mutación genética, como se aprecia en los gráficos de distribución presentados en el ítem 5 (Cáncer 80,7%/19,3%, Mutación genética 85,4%/14,6%).

Este tipo de desbalance no es una excepción, sino algo esperable en los datos médicos, y puede originarse en distintos niveles: en cómo se recogieron los datos (por ejemplo, la prevalencia real de la enfermedad en la población de origen), en el lugar de procedencia (zonas urbanas versus rurales, con distinto acceso a controles médicos), o en la forma en que se digitalizaron y depuraron los registros (Tasci et al., 2022). En nuestro caso, dado que el dataset es sintético, es probable que este desbalance simplemente refleje la proporción real esperada de casos positivos de cáncer y de mutaciones genéticas en la población general, más que un error de recolección.

Cuando esta diferencia es muy marcada, los modelos entrenados con esos datos tienden a aprender bien la clase mayoritaria, pero fallan en reconocer la minoritaria, generando muchos falsos negativos justamente en los casos que más interesa detectar (Gurcan & Soylu, 2024). Esto es especialmente riesgoso en nuestro contexto: dado que la gran mayoría de los pacientes no tiene cáncer ni mutación genética, un modelo entrenado sin corrección podría "aprender" a predecir casi siempre la clase negativa y aun así obtener una alta precisión general, mientras falla en detectar a las personas que sí tienen la enfermedad, que es precisamente el grupo que más importa identificar. En términos prácticos, si entrenáramos un modelo de predicción de cáncer con este dataset sin corregir el desbalance, probablemente tendría buen desempeño general (accuracy alto) pero mala sensibilidad clínica, que es justo lo que no se quiere en un modelo pensado para apoyar diagnósticos.

Frente a este problema, distintos estudios plantean que es necesario aplicar técnicas específicas de remuestreo antes de entrenar un modelo, de forma de equilibrar mejor la representación de cada clase, como el submuestreo (reducir la clase mayoritaria), el sobremuestreo (aumentar la clase minoritaria) o combinaciones de ambos enfoques (Gurcan & Soylu, 2024) (Tasci et al., 2022).
