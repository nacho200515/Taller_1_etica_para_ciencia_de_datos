"Not Applicable".
Nos genero curiosidad asi que la analizamos

```python
df[df['Breastfeeding_History'] == 'Not Applicable']['Gender'].value_counts()
```

```python
# cuantos hombres o mujeres hay
df['Gender'].value_counts()

# tabla cruzada: Breastfeeding_History segun su genero
df.groupby('Gender')['Breastfeeding_History'].value_counts()
```

```python
# Porcentaje de hombres con una respuesta si o no (que no deberia de)
hombres = df[df['Gender'] == 'Male']
pct_inconsistente = (hombres['Breastfeeding_History'] != 'Not Applicable').mean() * 100
print(f"{pct_inconsistente:.1f}% de los hombres tiene una respuesta que no les debería aplicar")
```

Al cruzar Breastfeeding_History con Gender, se observa que el 77% de los pacientes hombres presenta una respuesta "Yes" o "No" en una variable que biologicamente no les correspondería responder, y que la distribución de respuestas es prácticamente idéntica entre hombres y mujeres (~45% "Yes" en ambos grupos). Lo que genera una inconsistencia en los datos, sea probablemente a causa de que los datos fueron sinteticos.
